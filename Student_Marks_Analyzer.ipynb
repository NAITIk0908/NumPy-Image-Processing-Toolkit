#  Student Marks Analyzer


# 1 : Create subject names and marks


```python
import numpy as np

np.random.seed(42)

subjects = np.array(["Maths", "Physics", "Chemistry", "English", "Computer"])
marks = np.random.randint(35, 100, (50, 5))

print("Subjects:", subjects)
print("Marks created!")
```

# 2 :  Inspect shape, ndim, size and dtype

```python
print("Shape :", marks.shape)
print("ndim  :", marks.ndim)
print("Size  :", marks.size)
print("dtype :", marks.dtype)

print("\n--- Summary ---")
print(f"Students: {marks.shape[0]}, Subjects: {marks.shape[1]}")
print(f"Lowest mark : {marks.min()}")
print(f"Highest mark: {marks.max()}")
print(f"Overall mean: {marks.mean():.2f}")
```

# 3 : Slicing and indexing


```python
print("First 10 students:\n", marks[:10])
print("\nLast subject (negative index), first 10 students:", marks[:10, -1])
print("\nFirst student, all subjects:", marks[0])
print("\nFirst 3 subjects, first 5 students:\n", marks[:5, 0:3])
print("\nSubjects 2 to 4, students 10-14:\n", marks[10:15, 1:4])
```

 # 4 : Boolean masks




```python
below_40 = marks < 40
above_90 = marks > 90

print("Marks below 40 :", below_40.sum())
print("Marks above 90 :", above_90.sum())
print("Values above 90:", marks[above_90])

avg = marks.mean(axis=1)

band_A = avg >= 85
band_B = (avg >= 70) & (avg < 85)
band_C = (avg >= 55) & (avg < 70)
band_D = avg < 55

print("\nGrade bands (by student average):")
print("A (>=85)  :", band_A.sum(), "students ->", np.where(band_A)[0] + 1)
print("B (70-84) :", band_B.sum(), "students")
print("C (55-69) :", band_C.sum(), "students")
print("D (<55)   :", band_D.sum(), "students")
```

# 5 : Subject-wise toppers and pass percentage


```python
PASS_MARK = 40

print(f"{'Subject':<10}{'Average':>8}{'Topper(Student#)':>18}{'Top Marks':>11}{'Failed':>8}{'Pass %':>8}")
for i, sub in enumerate(subjects):
    col = marks[:, i]
    topper_idx = np.argmax(col)
    toppers_all = np.where(col == col.max())[0] + 1
    passed = col >= PASS_MARK
    pass_pct = passed.mean() * 100
    failed = (~passed).sum()
    print(f"{sub:<10}{col.mean():>8.2f}{topper_idx + 1:>18}{col.max():>11}{failed:>8}{pass_pct:>7.1f}%")
    if len(toppers_all) > 1:
        print(f"          (tied toppers: {toppers_all})")
```

# 6 : Overall toppers and failures

```python
total = marks.sum(axis=1)
print("Overall topper: Student", np.argmax(total) + 1, "with total", total.max())
print("Top 5 students:", np.argsort(total)[::-1][:5] + 1)

failed_students = np.where((marks < PASS_MARK).any(axis=1))[0] + 1
print("Students failing at least one subject:", failed_students)
print("Students passing all subjects:", (marks >= PASS_MARK).all(axis=1).sum())
```
